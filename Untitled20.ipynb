# Clustering Analysis on the Wine Dataset

This notebook demonstrates how to load a real-world dataset, preprocess it, apply clustering algorithms (K-Means and Agglomerative Clustering), evaluate their performance, and visualize the clusters.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.decomposition import PCA

# 1. Load the Wine dataset
wine = load_wine()
X = pd.DataFrame(wine.data, columns=wine.feature_names)
y = wine.target # Ground truth for comparison (optional)

print("Dataset Shape:", X.shape)
display(X.head())
```

### Data Preprocessing
Since clustering algorithms are sensitive to the scale of the features, we will standardize the dataset so that each feature has a mean of 0 and a variance of 1.

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

### Finding the Optimal Number of Clusters
We will use the **Elbow Method** (Inertia) and the **Silhouette Score** to find the optimal number of clusters ($K$) for K-Means.

```python
inertia = []
sil_scores = []
K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
    sil_scores.append(silhouette_score(X_scaled, kmeans.labels_))

# Plotting the Elbow Method and Silhouette Scores
fig, ax1 = plt.subplots(1, 2, figsize=(14, 5))

# Elbow plot
ax1[0].plot(K_range, inertia, 'bo-', label='Inertia')
ax1[0].set_xlabel('Number of Clusters (K)')
ax1[0].set_ylabel('Inertia (Within-cluster sum of squares)')
ax1[0].set_title('Elbow Method')
ax1[0].grid(True)

# Silhouette plot
ax1[1].plot(K_range, sil_scores, 'ro-', label='Silhouette Score')
ax1[1].set_xlabel('Number of Clusters (K)')
ax1[1].set_ylabel('Silhouette Score')
ax1[1].set_title('Silhouette Analysis')
ax1[1].grid(True)

plt.tight_layout()
plt.show()
```

### Model Training & Evaluation
We will fit both **K-Means** and **Agglomerative Clustering** with the chosen number of clusters ($K=3$) and compare their performances using **Silhouette Score** (higher is better) and **Davies-Bouldin Index** (lower is better).

```python
optimal_k = 3

# Train K-Means
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X_scaled)

# Train Agglomerative Clustering
agg = AgglomerativeClustering(n_clusters=optimal_k)
agg_labels = agg.fit_predict(X_scaled)

# Evaluation Metrics
metrics = {
    'Metric': ['Silhouette Score', 'Davies-Bouldin Index'],
    'K-Means': [
        silhouette_score(X_scaled, kmeans_labels),
        davies_bouldin_score(X_scaled, kmeans_labels)
    ],
    'Agglomerative': [
        silhouette_score(X_scaled, agg_labels),
        davies_bouldin_score(X_scaled, agg_labels)
    ]
}

metrics_df = pd.DataFrame(metrics)
display(metrics_df)
```

### Dimensionality Reduction and Visualization
We use PCA to reduce the features to 2 dimensions so we can visually inspect the cluster boundaries.

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Visualizing the cluster results side-by-side
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=kmeans_labels, palette='viridis', ax=axes[0])
axes[0].set_title('K-Means Clustering (K=3)')
axes[0].set_xlabel('PCA 1')
axes[0].set_ylabel('PCA 2')
axes[0].grid(True)

sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=agg_labels, palette='plasma', ax=axes[1])
axes[1].set_title('Agglomerative Clustering (K=3)')
axes[1].set_xlabel('PCA 1')
axes[1].set_ylabel('PCA 2')
axes[1].grid(True)

plt.tight_layout()
plt.show()
```
